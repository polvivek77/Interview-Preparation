```python
# 42. Lambda function
# A lambda function is a small anonymous function (function without a name).
# Used for:

# Short temporary functions
# Cleaner code
# Functional programming
# Mostly used with:
# map()
# filter()
# sorted()

addd = lambda a, b: a + b

addd(77,2)
```

```python
# Using Lambda with map()
numbers = [1, 2, 3, 4]

square = list(map(lambda x: x*x, numbers))

print(square)
```

```python
# Using Lambda with filter()
numbers = [1, 2, 3, 4, 5, 6]

even = list(filter(lambda x: x % 2 == 0, numbers))

print(even)
```

```python
# Using Lambda with sort()
employees = [
    ("John", 50000),
    ("Sam", 30000),
    ("Alex", 70000)
]

employees.sort(key=lambda x: x[1])

print(employees)
```

```python
# A generator is a special type of function that returns values one by one using the yield keyword instead of return.
# Generators save memory because they do not store all values at once.
def test():
    yield "A"
    yield "B"
    yield "C"

x = test()

print(next(x))
print(next(x))
print(next(x))
```

```python
# calculate total sales per region and per product category 
# data = { 'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'], 
# 'Product_Category': ['A', 'A', 'B', 'B', 'C', 'C', 'A', 'B'], 
# 'Sales': [100, 150, 200, 250, 300, 350, 400, 450] }

import pandas as pd

data = {
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Product_Category': ['A', 'A', 'B', 'B', 'C', 'C', 'A', 'B'],
    'Sales': [100, 150, 200, 250, 300, 350, 400, 450]
}

df = pd.DataFrame(data)

# Total Sales per Region
sales_per_region = df.groupby('Region')['Sales'].sum()

print("Sales Per Region:")
print(sales_per_region)

# Total Sales per Product Category
sales_per_category = df.groupby('Product_Category')['Sales'].sum()

print("\nSales Per Product Category:")
print(sales_per_category)
```

```python
def decorator_function(original_function):

    def wrapper_function(*args, **kwargs):
        print("Before the function is called")

        result = original_function(*args, **kwargs)

        print("After the function is called")

        return result

    return wrapper_function


@decorator_function
def my_function():
    print("Inside my_function")


my_function()
```

```python
# Matrix Multiplication
A = [
    [1, 2],
    [3, 4]
]

B = [
    [5, 6],
    [7, 8]
]

result = [
    [0, 0],
    [0, 0]
]

# Matrix multiplication
for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] += A[i][k] * B[k][j]

print(result)
```

```python
#31. Recursive factorial
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)

print(factorial(5))
```

```python
#32. Recursive Fibonacci
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)

print(fib(6))
```

```python
# find integers from given list using functioon and make a decorator function which will square the output numbers.
def decorator(func):
    def wrapper(g):
        result = func(g)
        return [x**2 for x in result]
    return wrapper

@decorator
def list_num(p):
    b=[]
    for i in p:
        if type(i)==int:
            b.append(i)
    return b

l = [1, 22, 3, 9, 'a', 'b', 'c']

v = list_num(l)
v
```
